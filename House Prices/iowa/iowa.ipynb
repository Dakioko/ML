```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
```

```python
#Load datasets
iowa_data = pd.read_csv("train.csv")
test_data = pd.read_csv("test.csv")
```

```python
iowa_data.dropna(axis=0)
#iowa_data.describe()
#iowa_data.info()
#iowa_data.columns
iowa_data
```

```python
# Feature Selection: Select X and y: X referes to a dataframe which includes the columns to be used for prediction. 
# y is the predictor target. The column we are predicting. Its a pandas series.

features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']
X = iowa_data[features]

y = iowa_data.SalePrice
```

```python
#Split data into train and test datasets. In this case, I will splist X, into _X_train,X-test,y_train, y_test,
X_train,X_test,y_train,y_test = train_test_split(X,y)
```

```python
#Define Model
iowa_model = DecisionTreeRegressor(random_state=1)

#Fit Model
iowa_model.fit(X_train,y_train)
```

```python
y_pred = iowa_model.predict(X_test)
```

```python
#Inspect your predictions and actual values from validation data.
print("Predictions")
#print(y_pred)
#print(y_pred.tolist().head())
print(iowa_model.predict(X_test.head()))
print(y_test.head())
```

```python
#Calculate the Mean Absolute Error in Validation Data
mean_absolute_error(y_test,y_pred)
```

```python
# Compare MAE scores from different values for max_leaf_nodes:

def get_mae(max_leaf_nodes):
    iowa_model2 = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes,random_state=0)
    iowa_model2.fit(X_train,y_train)
    y_pred= iowa_model2.predict(X_test)

    mae = mean_absolute_error(y_test,y_pred)
    return mae
```

```python
# compare MAE with differing values of max_leaf_nodes
print(f"Max Leaf Nodes: 5 \t {int(get_mae(5))}")
print(f"Max Leaf Nodes: 50 \t {int(get_mae(50))}")
print(f"Max Leaf Nodes: 500 \t {int(get_mae(500))}")
print(f"Max Leaf Nodes: 5000 \t {int(get_mae(5000))}")
print(f"Max Leaf Nodes: 50000 \t {int(get_mae(50000))}")
print(f"Max Leaf Nodes: 500000 \t {int(get_mae(500000))}")
```

```python
# compare MAE with differing values of max_leaf_nodes using a for Loop
for max_leaf_nodes in [5, 50, 500, 5000,100]:
    my_mae = get_mae(max_leaf_nodes)
    #print("Max leaf nodes: %d  \t\t Mean Absolute Error:  %d" %(max_leaf_nodes, my_mae))
    print(f"Max leaf nodes: {max_leaf_nodes} \t\t Mean Absolute Error: {int(my_mae)}")
```

```python
#Write a loop that tries the following values for max_leaf_nodes from a set of possible values.

#Call the get_mae function on each value of max_leaf_nodes. Store the output in some way that allows you to select the value of max_leaf_nodes that gives the most accurate model on your data.

candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500,5000]
# Write loop to find the ideal tree size from candidate_max_leaf_nodes
for max_leaf_nodes in [5, 25, 50, 100, 250, 500]:
    my_mae = get_mae(max_leaf_nodes)
   
# Store the best value of max_leaf_nodes (it will be either 5, 25, 50, 100, 250 or 500)

scores = {leaf_size: get_mae(leaf_size) for leaf_size in candidate_max_leaf_nodes}
best_tree_size = min(scores, key=scores.get)
best_tree_size
```

```python
# Generate a list of all values between 1 and 1000 (excluding 0, as it wouldn't make sense for max_leaf_nodes)
# Define the range of candidate max_leaf_nodes
# Define the range of candidate max_leaf_nodes
candidate_max_leaf_nodes = range(2, 1001)  # All values from 2 to 1000

# Write a loop to find the ideal tree size from candidate_max_leaf_nodes
scores = {}
for max_leaf_nodes in candidate_max_leaf_nodes:
    scores[max_leaf_nodes] = get_mae(max_leaf_nodes)

# Store the best value of max_leaf_nodes
best_tree_size = min(scores, key=scores.get)

# Output the best tree size
print(f"The ideal max_leaf_nodes is: {best_tree_size}")
```

#### Fit Model Using All Data
> You know the best tree size. If you were going to deploy this model in practice, you would make it even more accurate by using all of the data and keeping that tree size. That is, you don't need to hold out the validation data now that you've made all your modeling decisions.

```python
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size, random_state=1)

# fit the final model
final_model.fit(X, y)
```

```python
#Predictions using the final model.
y_pred_final = final_model.predict(X_test)
pred_5 = final_model.predict(X_test.head())
print(pred_5)
print(y_test.head())
```
