```python
#Import necessary libraries. 
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import mean_absolute_error
```

```python
train_data = pd.read_csv('train.csv',index_col='PassengerId')
train_data.head()
```

```python
train_data.columns
```

```python
test_data = pd.read_csv('test.csv').set_index('PassengerId')
test_data.head()
#print(test_data.head().to_string())
```

```python
# Select the Target Variable(y) and the Features (Independent Variables)(X)
y = train_data.Survived
features = ['Pclass','Sex','Fare','Age','SibSp','Parch','Embarked']
features
X = pd.get_dummies(train_data[features])
X
```

```python
#Split the Train Dataset into Train and Test/Validation Datasets.
X_train,X_test,y_train,y_test = train_test_split(X,y)
```

```python
# Model Selection
# I will use the Random Forest Classification

model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
#Fit Model
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
#Check accuracy of the model

from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy Score: {accuracy * 100:.2f}%')
```
