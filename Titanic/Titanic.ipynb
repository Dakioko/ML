```python
#Get the total number of women
women = titanic_data.loc[titanic_data.Sex == 'female']
total_women = len(women)
print("Total number of Women:", total_women)
surived_women = sum(women.Survived)
print("Total number of women who survived:", surived_women)


rate_women = surived_women/total_women
print("% of women who survived:", rate_women)
```

```python
# Number of total passengers
total = len(titanic_data)
print(f"Total Passengers: {total}")

survived = (titanic_data.Survived==1).sum()
print(f"Survived Passengers: {survived}")

dead = (titanic_data.Survived==0).sum()
print(f"Dead Passengers: {dead}")

minors = (titanic_data.Age<18).sum()
print(f"Minors: {minors}")
```

```python
survived_fraction = survived/total
print(f"Survived Fraction: {survived_fraction}")

minors_fraction = minors/total
print(f"Minors Fraction: {minors_fraction}")
```
